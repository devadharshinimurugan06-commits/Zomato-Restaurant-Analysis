## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

reviews_clean = reviews_df.copy()
metadata_clean = metadata_df.copy()

# Remove duplicate rows
reviews_clean.drop_duplicates(inplace=True)
metadata_clean.drop_duplicates(inplace=True)

# Handle missing values - Reviews Dataset
reviews_clean['Reviewer'] = reviews_clean['Reviewer'].fillna('Unknown')
reviews_clean['Review'] = reviews_clean['Review'].fillna('')
reviews_clean['Rating'] = reviews_clean['Rating'].fillna('Unknown')
reviews_clean['Metadata'] = reviews_clean['Metadata'].fillna('Unknown')
reviews_clean['Time'] = reviews_clean['Time'].fillna('Unknown')

# Handle missing values - Metadata Dataset
metadata_clean['Collections'] = metadata_clean['Collections'].fillna('Not Available')
metadata_clean['Timings'] = metadata_clean['Timings'].fillna('Not Available')

# Remove extra spaces from text columns
for col in reviews_clean.select_dtypes(include='object').columns:
    reviews_clean[col] = reviews_clean[col].str.strip()

for col in metadata_clean.select_dtypes(include='object').columns:
    metadata_clean[col] = metadata_clean[col].str.strip()

print("Data Wrangling Completed Successfully!")

print("\nReviews Dataset Shape:", reviews_clean.shape)
print("Metadata Dataset Shape:", metadata_clean.shape)
````

````python
# Merge Dataset

zomato_df = pd.merge(
    reviews_clean,
    metadata_clean,
    left_on='Restaurant',
    right_on='Name',
    how='left'
)

print("Datasets Merged Successfully!")
print("Merged Dataset Shape:", zomato_df.shape)

display(zomato_df.head())
````

````python
# Save Merged Dataset

merged_path = dataset_path + "/zomato_merged_dataset.csv"

zomato_df.to_csv(merged_path, index=False)

print("Merged Dataset Saved Successfully!")
print("File Location:", merged_path)
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis :

There is no significant relationship between the restaurant Cost and Rating.

Alternate Hypothesis :

There is a significant relationship between the restaurant Cost and Rating.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import pearsonr

# Convert Cost to numeric
zomato_df['Cost'] = (
    zomato_df['Cost']
    .astype(str)
    .str.replace(',', '', regex=False)
)

zomato_df['Cost'] = pd.to_numeric(zomato_df['Cost'], errors='coerce')
zomato_df['Rating'] = pd.to_numeric(zomato_df['Rating'], errors='coerce')

df1 = zomato_df[['Cost','Rating']].dropna()

corr, p_value = pearsonr(df1['Cost'], df1['Rating'])

print("Correlation :", corr)
print("P-value :", p_value)

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test

##### Why did you choose the specific statistical test?

Pearson Correlation Test is suitable because both Cost and Rating are numerical variables. It measures the strength of the linear relationship between them and provides a p-value to determine statistical significance.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis :

There is no significant difference in restaurant Cost among different Cuisine categories.

Alternate Hypothesis :

There is a significant difference in restaurant Cost among different Cuisine categories.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import f_oneway

zomato_df['Cost'] = (
    zomato_df['Cost']
    .astype(str)
    .str.replace(',', '', regex=False)
)
zomato_df['Cost'] = pd.to_numeric(zomato_df['Cost'], errors='coerce')

top_cuisines = zomato_df['Cuisines'].value_counts().head(5).index

groups = []

for cuisine in top_cuisines:
    groups.append(
        zomato_df[zomato_df['Cuisines']==cuisine]['Cost'].dropna()
    )

f_stat, p_value = f_oneway(*groups)

print("F Statistic :", f_stat)
print("P-value :", p_value)

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA Test

##### Why did you choose the specific statistical test?

One-Way ANOVA is appropriate because it compares the mean Cost across more than two cuisine groups and determines whether the observed differences are statistically significant.

### 2. Handling Outliers

````python

# Convert Cost into numeric

zomato_df['Cost'] = (
    zomato_df['Cost']
    .astype(str)
    .str.replace(',', '', regex=False)
)

zomato_df['Cost'] = pd.to_numeric(zomato_df['Cost'], errors='coerce')

# Detect Outliers using IQR

Q1 = zomato_df['Cost'].quantile(0.25)
Q3 = zomato_df['Cost'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

zomato_df = zomato_df[
    (zomato_df['Cost'] >= lower) &
    (zomato_df['Cost'] <= upper)
]

print("Dataset Shape after Outlier Treatment:", zomato_df.shape)
print("After Outlier Removal :", zomato_df.shape)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

The Interquartile Range IQR method was used to detect and remove outliers from the Cost column. Values outside the lower and upper IQR limits were treated as outliers and removed. This technique is robust because it is less affected by extreme values and helps improve the quality and reliability of statistical analysis and machine learning models.

### 4. Textual Data Preprocessing
(It's mandatory for textual dataset i.e., NLP, Sentiment Analysis, Text Clustering etc.)

#### 1. Expand Contraction

````python
# Expand Contraction
import re

contractions = {
    "can't": "cannot",
    "won't": "will not",
    "n't": " not",
    "'re": " are",
    "'s": " is",
    "'d": " would",
    "'ll": " will",
    "'ve": " have",
    "'m": " am"
}

def expand_contractions(text):
    text = str(text)
    for key, value in contractions.items():
        text = re.sub(key, value, text)
    return text

zomato_df['Review'] = zomato_df['Review'].apply(expand_contractions)

print(zomato_df['Review'].head())
````

#### 2. Lower Casing

````python
# Lower Casing
zomato_df['Review'] = zomato_df['Review'].str.lower()

print(zomato_df['Review'].head())
````

#### 3. Removing Punctuations

````python
# Remove Punctuations
import string

zomato_df['Review'] = zomato_df['Review'].str.translate(
    str.maketrans('', '', string.punctuation)
)

print(zomato_df['Review'].head())
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove words and digits contain digits
import re

zomato_df['Review'] = zomato_df['Review'].apply(
    lambda x: re.sub(r'http\S+|www\S+', '', str(x))
)

zomato_df['Review'] = zomato_df['Review'].apply(
    lambda x: re.sub(r'\w*\d\w*', '', str(x))
)

print(zomato_df['Review'].head())
````

````python
import re

zomato_df['Review'] = zomato_df['Review'].apply(
    lambda x: re.sub(r'[^\w\s]', ' ', str(x))
)

zomato_df['Review'] = zomato_df['Review'].str.replace(r'\s+', ' ', regex=True).str.strip()
````

## ***7. ML Model Implementation***

### ML Model - 1

````python
# ML Model - 1 Implementation

# Fit the Algorithm

# Predict on the model

from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

# Create Model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Fit the Algorithm
kmeans.fit(X_scaled)

# Predict on the model
kmeans_labels = kmeans.predict(X_scaled)

# Store Cluster Labels
zomato_df["KMeans_Cluster"] = kmeans_labels

print(zomato_df[["KMeans_Cluster"]].head())

````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
# Evaluation Metrics

kmeans_sil = silhouette_score(
    X_scaled,
    kmeans_labels
)

kmeans_db = davies_bouldin_score(
    X_scaled,
    kmeans_labels
)

kmeans_ch = calinski_harabasz_score(
    X_scaled,
    kmeans_labels
)

print("Silhouette Score :", kmeans_sil)
print("Davies-Bouldin Index :", kmeans_db)
print("Calinski-Harabasz Score :", kmeans_ch)

# Visualizing evaluation Metric Score chart

import matplotlib.pyplot as plt

metrics = [
    "Silhouette",
    "Davies-Bouldin",
    "Calinski-Harabasz"
]

scores = [
    kmeans_sil,
    kmeans_db,
    kmeans_ch
]

plt.figure(figsize=(7,5))

plt.bar(metrics, scores)

plt.title("K-Means Evaluation Metrics")

plt.ylabel("Score")

plt.show()
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
#  Implementation with hyperparameter optimization techniques

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np

silhouette_scores = []

k_values = range(2,11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)

    labels = model.predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(score)

# Best K
best_k = k_values[np.argmax(silhouette_scores)]

print("Best Number of Clusters :", best_k)

# Best Model
best_model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

# Fit the Algorithm
best_model.fit(X_scaled)

# Predict on the model
best_labels = best_model.predict(X_scaled)

# Save Cluster Labels
zomato_df["Optimized_KMeans_Cluster"] = best_labels

# Evaluation Metrics
best_silhouette = silhouette_score(
    X_scaled,
    best_labels
)

best_davies = davies_bouldin_score(
    X_scaled,
    best_labels
)

best_calinski = calinski_harabasz_score(
    X_scaled,
    best_labels
)

print("\nOptimized Model Evaluation")
print("Silhouette Score :", best_silhouette)
print("Davies-Bouldin Index :", best_davies)
print("Calinski-Harabasz Score :", best_calinski)
````

##### Which hyperparameter optimization technique have you used and why?

Hyperparameter optimization was performed using Silhouette Score-based search. Different values of K (2 to 10) were evaluated, and the number of clusters with the highest Silhouette Score was selected as the optimal value because it produces compact and well-separated clusters.Cross-validation was not performed because K-Means is an unsupervised learning algorithm and does not use labeled target variables.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

Yes. After selecting the optimal number of clusters based on the Silhouette Score, the clustering performance improved. The optimized K-Means model produced a better Silhouette Score, a lower Davies–Bouldin Index, and a higher Calinski–Harabasz Score compared to the initial K-Means model, indicating better cluster quality.
